```python
from langchain_groq import ChatGroq
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.runnables import chain

# the building blocks
template = ChatPromptTemplate.from_messages([
    ('system', 'You are a helpful assistant.'),
    ('human', '{question}')
])

model = ChatGroq(model="openai/gpt-oss-120b")

# Combine them in a function
# @chain decorator adds the same Runnable interface for any function you write

@chain
def chatbot(values):
    prompt = template.invoke(values)
    return model.invoke(prompt)

# Use it
chatbot.invoke({"question": "Which model providers offer LLMs?"})
```

```python
@chain
def chatbot(values):
    prompt = template.invoke(values)
    
    for token in model.stream(prompt):
        yield token # yield turns a standard function into a generator
                    # return: Ends the function completely and destroys its local variables.
                    # yield: Pauses the function, saves its state, and yields a single value [1]. The function resumes right after the yield statement when called again

for part in chatbot.stream({
    "question":"Which model providers offer LLMs?"
}):
    print(part)
```

```python
@chain
async def chatbot(values):
    prompt = await template.ainvoke(values)
    return await model.ainvoke(prompt)

await chatbot.ainvoke({"question":"Which model providers offer LLMs?"})
```

### Declarative Composition

```python
from langchain_groq import ChatGroq
from langchain_core.prompts import ChatPromptTemplate

#the building blocks

template = ChatPromptTemplate.from_messages([
    ('system', 'You are a helpful assistant.'),
    ('human', '{question}'),
])

model = ChatGroq(model="openai/gpt-oss-120b")

# Combine them with the | operator
chatbot = template | model

# Use it
chatbot.invoke({"question": "Which model providers offer LLMs?"})
```

```python
chatbot = template | model

for part in chatbot.stream({
    "question":"Which model providers offer LLMs?"
}):
    print(part)
```

```python
chatbot = template | model

await chatbot.ainvoke({
    "question": "Which model providers offer LLMs?"
})
```
